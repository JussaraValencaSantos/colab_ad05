# 🐱 Versão 1: uma imagem de gato

Faz uma requisição à API [The Cat API](https://thecatapi.com/), lê o endereço da imagem na resposta JSON, baixa e salva em `imagens/cat/cat.jpg`.

> No Colab os arquivos ficam em `/content` e **são apagados quando a sessão termina**. Baixe o que quiser guardar pelo painel de arquivos (ícone de pasta, à esquerda).

## 1. Importações

```python
import os
import requests
from IPython.display import Image, display
```

## 2. Pasta de saída

```python
pasta = os.path.join(os.getcwd(), 'imagens/cat')
os.makedirs(pasta, exist_ok=True)
caminho = os.path.join(pasta, 'cat.jpg')
```

## 3. Requisição à API e download da imagem

```python
url = 'https://api.thecatapi.com/v1/images/search'
response = requests.get(url)
data = response.json()
img_url = data[0]['url']
img_data = requests.get(img_url).content
with open(caminho, 'wb') as handler:
    handler.write(img_data)
print(f"Imagem do gato salva em {caminho}!")
```

## 4. Visualizar a imagem

```python
display(Image(filename=caminho, width=400))
```
